```python
def freqTable(columnName,dataset):
        freqTable=pd.DataFrame(columns=["Unique_values","Frequency","Relative_frequency","Cumsum"])
        freqTable["Unique_values"]=dataset[columnName].value_counts().index
        freqTable["Frequency"]=dataset[columnName].value_counts().values
        freqTable["Relative_frequency"]=(freqTable["Frequency"]/4493)
        freqTable["Cumsum"]=freqTable["Relative_frequency"].cumsum()
        return freqTable
freqTable(columnName,dataset)
```

```python
import time
import numpy as np
from sklearn.feature_selection import SelectKBest
from sklearn.feature_selection import chi2
from sklearn.feature_selection import RFE
from sklearn.linear_model import LogisticRegression
import pickle
import matplotlib.pyplot as plt

# train and test set split before selectK-best
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(independent, dependent, test_size=0.30, random_state=42)

def selectkbest(x_train,y_train,x_test,n):
        test = SelectKBest(score_func=chi2, k=n)
        x_train_selected= test.fit_transform(x_train,y_train)
        x_test_selected = test.transform(x_test)
        return x_train_selected,x_test_selected,test

x_train_selected,x_test_selected,selector_obj = selectkbest(x_train,y_train,x_test,2)

def cm_prediction(classifier,x_test_selected):
     y_pred = classifier.predict(x_test_selected)
        
        # Making the Confusion Matrix
     from sklearn.metrics import confusion_matrix
     cm = confusion_matrix(y_test, y_pred)
        
     from sklearn.metrics import accuracy_score 
     from sklearn.metrics import classification_report 
        #from sklearn.metrics import confusion_matrix
        #cm = confusion_matrix(y_test, y_pred)
        
     Accuracy=accuracy_score(y_test, y_pred )
        
     report=classification_report(y_test, y_pred)
     return  classifier,Accuracy,report,x_test_selected,y_test,cm

def logistic(x_train_selected,y_train,x_test_selected):       
        # Fitting K-NN to the Training set
        from sklearn.linear_model import LogisticRegression
        classifier = LogisticRegression(random_state = 0)
        classifier.fit(x_train_selected, y_train)
        classifier,Accuracy,report,x_test_selected,y_test,cm=cm_prediction(classifier,x_test_selected)
        return  classifier,Accuracy,report,x_test_selected,y_test,cm      
    
def svm_linear(x_train_selected,y_train,x_test_selected):
                
        from sklearn.svm import SVC
        classifier = SVC(kernel = 'linear', random_state = 0)
        classifier.fit(x_train_selected, y_train)
        classifier,Accuracy,report,x_test_selected,y_test,cm=cm_prediction(classifier,x_test_selected)
        return  classifier,Accuracy,report,x_test_selected,y_test,cm
    
def svm_NL(x_train_selected,y_train,x_test_selected):
                
        from sklearn.svm import SVC
        classifier = SVC(kernel = 'rbf', random_state = 0)
        classifier.fit(x_train_selected, y_train)
        classifier,Accuracy,report,x_test_selected,y_test,cm=cm_prediction(classifier,x_test_selected)
        return  classifier,Accuracy,report,x_test_selected,y_test,cm
   
def Navie(x_train_selected,y_train,x_test_selected):       
        # Fitting K-NN to the Training set
        from sklearn.naive_bayes import GaussianNB
        classifier = GaussianNB()
        classifier.fit(x_train_selected, y_train)
        classifier,Accuracy,report,x_test_selected,y_test,cm=cm_prediction(classifier,x_test_selected)
        return  classifier,Accuracy,report,x_test_selected,y_test,cm         
    
def knn(x_train_selected,y_train,x_test_selected):
           
        # Fitting K-NN to the Training set
        from sklearn.neighbors import KNeighborsClassifier
        classifier = KNeighborsClassifier(n_neighbors = 5, metric = 'minkowski', p = 2)
        classifier.fit(x_train_selected, y_train)
        classifier,Accuracy,report,x_test_selected,y_test,cm=cm_prediction(classifier,x_test_selected)
        return  classifier,Accuracy,report,x_test_selected,y_test,cm
def Decision(x_train_selected,y_train,x_test_selected):
        
        # Fitting K-NN to the Training set
        from sklearn.tree import DecisionTreeClassifier
        classifier = DecisionTreeClassifier(criterion = 'entropy', random_state = 0)
        classifier.fit(x_train_selected, y_train)
        classifier,Accuracy,report,x_test_selected,y_test,cm=cm_prediction(classifier,x_test_selected)
        return  classifier,Accuracy,report,x_test_selected,y_test,cm      


def random(x_train_selected,y_train,x_test_selected):
        
        # Fitting K-NN to the Training set
        from sklearn.ensemble import RandomForestClassifier
        classifier = RandomForestClassifier(n_estimators = 10, criterion = 'entropy', random_state = 0)
        classifier.fit(x_train_selected, y_train)
        classifier,Accuracy,report,X_test_selected,y_test,cm=cm_prediction(classifier,x_test_selected)
        return  classifier,Accuracy,report,x_test_selected,y_test,cm
    
def selectk_Classification(acclog,accsvml,accsvmnl,accknn,accnav,accdes,accrf): 
    
    dataframe=pd.DataFrame(index=['ChiSquare'],columns=['Logistic','SVMl','SVMnl','KNN','Navie','Decision','Random'])
    for number,idex in enumerate(dataframe.index):      
        dataframe['Logistic'][idex]=acclog[number]       
        dataframe['SVMl'][idex]=accsvml[number]
        dataframe['SVMnl'][idex]=accsvmnl[number]
        dataframe['KNN'][idex]=accknn[number]
        dataframe['Navie'][idex]=accnav[number]
        dataframe['Decision'][idex]=accdes[number]
        dataframe['Random'][idex]=accrf[number]
    return dataframe
    
```

```python
classifier,Accuracy,report,x_test_selected,y_test,cm=logistic(x_train_selected,y_train,x_test_selected)
acclog.append(Accuracy)

classifier,Accuracy,report,x_test_selected,y_test,cm=svm_linear(x_train_selected,y_train,x_test_selected)  
accsvml.append(Accuracy)
    
classifier,Accuracy,report,x_test_selected,y_test,cm=svm_NL(x_train_selected,y_train,x_test_selected)  
accsvmnl.append(Accuracy)
    
classifier,Accuracy,report,x_test_selected,y_test,cm=knn(x_train_selected,y_train,x_test_selected)  
accknn.append(Accuracy)
    
classifier,Accuracy,report,x_test_selected,y_test,cm=Navie(x_train_selected,y_train,x_test_selected)  
accnav.append(Accuracy)
    
classifier,Accuracy,report,x_test_selected,y_test,cm=Decision(x_train_selected,y_train,x_test_selected)  
accdes.append(Accuracy)
    
classifier,Accuracy,report,x_test_selected,y_test,cm=random(x_train_selected,y_train,x_test_selected)  
accrf.append(Accuracy)
    
result=selectk_Classification(acclog,accsvml,accsvmnl,accknn,accnav,accdes,accrf)
```
